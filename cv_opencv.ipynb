```python
from google.colab.patches import cv2_imshow

img = cv.imread("/content/dog.jpg")
cv2_imshow(img)
```

```python
print("shape:", img.shape)
```

```python
print("height:", img.shape[0])
print("width:", img.shape[1])
print("channels:", img.shape[2])
```

```python
(b, g, r) = img[1,1]
print("pixel at (1,1) - Red:", r, "Green:", g, "Blue:", b )
```

```python
img_in_gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
cv2_imshow(img_in_gray)
```

```python
img_resized = cv.resize(img_in_gray, (270, 200))
cv2_imshow(img_resized)
```

```python
cropped = img_in_gray[0:200, 0:300]
cv2_imshow(cropped)
```

```python
img_rotated = cv.rotate(cropped, cv.ROTATE_90_CLOCKWISE)
cv2_imshow(img_rotated)
```

```python
img_flipped = cv.flip(img_rotated, 1)
cv2_imshow(img_flipped)
```

```python
cv.imwrite("dog_in_gray.png", img_in_gray)
```

# open cv video streaming **codes**

```python
print("hello world")

import cv2 as cv

cap = cv.VideoCapture(0, cv.CAP_AVFOUNDATION)

face_cascade = cv.CascadeClassifier(cv.data.haarcascades + "haarcascade_frontalface_default.xml")

while True:
    success, frame = cap.read() # True / False

    if not success: break

    faces = face_cascade.detectMultiScale(frame, 1.5, 4)
    # frame_gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)


    for (x, y, w, h) in faces:
        cv.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    cv.imshow("Real-time Face Detection", frame)

    # cv.imshow('Original', frame_gray)

    if(cv.waitKey(1) & 0xFF == ord('q')): break

cap.release()
cv.destroyAllWindows()
```
